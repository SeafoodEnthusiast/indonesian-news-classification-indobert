```python
# Text Preprocessing

import pandas as pd
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

# Load Data
df = pd.read_csv("dataset_berita_100.csv")


stop_factory = StopWordRemoverFactory()
stopwords = stop_factory.get_stop_words()

stem_factory = StemmerFactory()
stemmer = stem_factory.create_stemmer()

def preprocess_text(text):
    # 1. Cleansing
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)

    # 2. Tokenization & Filtering
    words = text.split()
    words = [w for w in words if w not in stopwords]

    # 3. Stemming
    clean_text = stemmer.stem(" ".join(words))

    return clean_text


df['Clean_Teks'] = df['Teks'].apply(preprocess_text)

# Result
print("Hasil Preprocessing:")
print(df[['Teks', 'Clean_Teks']].head())

df.to_csv("dataset_berita_100_clean.csv", index=False)
```

#1C) LLM Model Classification

Pada tahap ini, Saya menggunakan IndoBERT (indobenchmark/indobert-base-p1), sebuah model Large Language Model (LLM) berbasis arsitektur BERT yang telah dilatih (pretrained) menggunakan dataset masif Bahasa Indonesia (Indo4B). Penggunaan model ini dipilih karena kemampuannya memahami konteks semantik bahasa Indonesia jauh lebih baik daripada model multibahasa biasa.

Hyperparameter Tuning: Parameter Tuning dilakukan pada 2 Parameter Utama

Epochs (10): Dinaikkan dari standar (3) menjadi 10 agar model bisa belajar lebih mendalam mengingat dataset yang kecil.


Learning Rate (5e-5): Diatur agar proses update bobot model berjalan optimal.

```python
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
from transformers import BertTokenizer, BertForSequenceClassification, Trainer, TrainingArguments

# Setup Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device yang digunakan: {device}")


df = pd.read_csv("dataset_berita_100_clean.csv")


model_name = "indobenchmark/indobert-base-p1"
tokenizer = BertTokenizer.from_pretrained(model_name)

class NewsDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)

# Fungsi Metrik Evaluasi
def compute_metrics(pred):
    labels = pred.label_ids
    preds = pred.predictions.argmax(-1)
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='weighted', zero_division=0)
    acc = accuracy_score(labels, preds)
    return {'accuracy': acc, 'f1': f1, 'precision': precision, 'recall': recall}

# Fungsi Untuk Training
def train_model(texts, labels, num_labels, output_dir):
    # Split Data (80% Train, 20% Test)
    train_texts, val_texts, train_labels, val_labels = train_test_split(texts, labels, test_size=0.2, random_state=42)

    # Tokenisasi
    train_encodings = tokenizer(train_texts.tolist(), truncation=True, padding=True, max_length=128)
    val_encodings = tokenizer(val_texts.tolist(), truncation=True, padding=True, max_length=128)

    # Buat Dataset Object
    train_dataset = NewsDataset(train_encodings, train_labels.tolist())
    val_dataset = NewsDataset(val_encodings, val_labels.tolist())

    # Load Pretrained IndoBERT
    model = BertForSequenceClassification.from_pretrained(model_name, num_labels=num_labels)
    model.to(device)

    # HYPERPARAMETER TUNING
    training_args = TrainingArguments(
        output_dir=output_dir,
        num_train_epochs=10,
        learning_rate=5e-5,
        per_device_train_batch_size=8,
        eval_strategy="epoch",
        save_strategy="no",
        logging_dir=f'{output_dir}/logs',
        logging_steps=10,
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_dataset,
        eval_dataset=val_dataset,
        compute_metrics=compute_metrics
    )

    print(f"Mulai Training {output_dir}...")
    trainer.train()
    return trainer.evaluate(), trainer
```

```python
# MODEL 1: 1-STAGE (Langsung 5 Kelas)

print("MODEL 1-STAGE (5 Label)")

# Mapping Label ke Angka (0-4)
label_map_1s = {
    'Liga Inggris': 0, 'Liga Indonesia': 1, 'Liga Spanyol': 2,
    'Liga Italia': 3, 'Non Sepak Bola': 4
}
df['label_1s'] = df['Label'].map(label_map_1s)

# Train Model
results_1s, trainer_1s = train_model(df['Clean_Teks'], df['label_1s'], num_labels=5, output_dir='./model_1stage')
print("Hasil Evaluasi 1-Stage:", results_1s)
```

```python
# MODEL 2: 2-STAGE (Step 1: Bola vs Non-Bola, Step 2: Liga)
print("MODEL 2-STAGE")

# STAGE 1: Klasifikasi Biner (Sepak Bola vs Non Sepak Bola)
print("[Stage 1] Training Model Biner (Bola vs Non-Bola)...")
# Label: 0 = Non Bola, 1 = Bola
df['label_s1'] = df['Label'].apply(lambda x: 0 if x == 'Non Sepak Bola' else 1)

results_s1, trainer_s1 = train_model(df['Clean_Teks'], df['label_s1'], num_labels=2, output_dir='./model_2stage_binary')
print("Hasil Stage 1 (Binary):", results_s1)


# STAGE 2: Klasifikasi Liga (Hanya Data Sepak Bola)
print("[Stage 2] Training Model Liga (Khusus Sepak Bola)...")
# Filter data: Ambil hanya yang Sepak Bola
df_bola = df[df['Label'] != 'Non Sepak Bola'].copy()

# Mapping Label Liga (0-3)
label_map_s2 = {
    'Liga Inggris': 0, 'Liga Indonesia': 1, 'Liga Spanyol': 2, 'Liga Italia': 3
}
df_bola['label_s2'] = df_bola['Label'].map(label_map_s2)

results_s2, trainer_s2 = train_model(df_bola['Clean_Teks'], df_bola['label_s2'], num_labels=4, output_dir='./model_2stage_league')
print("Hasil Stage 2 (League):", results_s2)
```

#1D) Model Comparison

```python
import pandas as pd

# Daftar model
models_to_check = [
    ("1-Stage (All Class)", trainer_1s),
    ("2-Stage (Step 1: Biner)", trainer_s1),
    ("2-Stage (Step 2: League)", trainer_s2)
]

hasil_akhir = []

for nama_model, trainer in models_to_check:
    train_metrics = trainer.predict(trainer.train_dataset).metrics
    hasil_akhir.append({
        "Model": nama_model,
        "Dataset": "Training",
        "Accuracy": train_metrics['test_accuracy'],
        "F1-Score": train_metrics['test_f1']
    })

    test_metrics = trainer.evaluate()
    hasil_akhir.append({
        "Model": nama_model,
        "Dataset": "Testing",
        "Accuracy": test_metrics['eval_accuracy'],
        "F1-Score": test_metrics['eval_f1']
    })

df_final = pd.DataFrame(hasil_akhir)
print("Tabel Hasil Evaluasi")
display(df_final)
```

Analisis Evaluasi: Evaluasi dilakukan dengan membagi data menjadi 80% Training dan 20% Testing. Berdasarkan tabel perbandingan didapat hasil sebagai berikut:



Performa Training: Seluruh model mencapai akurasi 1.0 (100%) pada data training, menunjukkan model berhasil mempelajari pola data dengan sempurna.



Performa Testing:



Model 1-Stage unggul dengan akurasi 90%.



Model 2-Stage memiliki performa gabungan sebesar 87.5%.



Kesimpulan: Pendekatan 1-Stage terbukti lebih efektif dan efisien untuk kasus ini. Model 2-Stage memiliki risiko error propagation (kesalahan berantai), di mana jika prediksi Biner (Bola/Non-Bola) salah, maka prediksi Liga pasti akan salah. Oleh karena itu, arsitektur 1-Stage lebih disarankan untuk kasus ini.
